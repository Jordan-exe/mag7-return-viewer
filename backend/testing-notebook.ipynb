```python
from datetime import UTC, datetime, timedelta

import yfinance as yf
```

```python
m7 = ["MSFT", "AAPL", "GOOGL", "AMZN", "NVDA", "META", "TSLA", "UKNOWN_TICKER"]
start = (
    datetime(year=2026, month=9, day=1, tzinfo=UTC).date() - timedelta(days=10)
).isoformat()
end = (datetime.now(tz=UTC).date() + timedelta(days=1)).isoformat()
data = yf.download(m7, start=start, end=end, auto_adjust=True)
closes = data.Close
```
